```python
prodotti = {
    'roots': {
        'prod1': {'name': 'mandioca', 'price': 15.0, 'quantity': 20, 'weight': 0.1}
    },
    'fruits': {
        'prod2': {'name': 'banana', 'price': 5.0, 'quantity': 100, 'weight': 0.5}
    },
    'meat': {
        'prod3': {'name': 'picanha', 'price': 30.0, 'quantity': 8, 'weight': 1.0},
        'prod4': {'name': 'fiorentina', 'price': 35.0, 'quantity': 10, 'weight': 2.0}
    }
}

orders_track = []


# Student a

def show_all():
    for c, items in prodotti.items():
        print(f"{c}")
        for id, e in items.items():
            print(f"  {id}, {e['name']}, euro{e['price']}, stock: {e['quantity']}")

def show_by_category(cat):
    if cat not in prodotti:
        print("product not found")
        return
    for id, e in prodotti[cat].items():
        print(f"{id}, {e['name']}, {e['price']}, qty: {e['quantity']}")

def set_price(name, price):
    for c in prodotti.values():
        for e in c.values():
            if e['name'] == name:
                e['price'] = float(price)
                return
    print("not found")

def set_stock(name, qty):
    for c in prodotti.values():
        for e in c.values():
            if e['name'] == name:
                e['quantity'] = int(qty)
                return
    print("not found")


# student b

def validate_stock(prod, q):
    for c in prodotti.values():
        for e in c.values():
            if e['name'] == prod:
                return e['quantity'] >= q
    return False

def apply_coupon(code, tot):
    if code == 'nik05':
        return max(0, tot - 5)
    elif code == 'falco10':
        return tot * 0.9
    return tot

def place_order(product, qty):
    if not validate_stock(product, qty):
        print("not enough quantity")
        return

    for cat_name, c in prodotti.items():
        for e in c.values():
            if e['name'] == product:
                sub = e['price'] * qty
                w = e['weight'] * qty

                if sub > 100:
                    ship = 0
                elif w > 5:
                    ship = 8
                else:
                    ship = 5

                code = input("coupon code, bianco= skip: ")
                sub = apply_coupon(code, sub)
                tot = sub + ship

                o = {
                    'name': product,
                    'category': cat_name,
                    'quantity': qty,
                    'price': tot,
                    'status': 'completed'
                }
                e['quantity'] -= qty
                orders_track.append(o)
                print(f"order placed. subtotal: euro{sub}, shipping: euro{ship}, total: euro{tot}")
                return


# c

def order_summary():
    if not orders_track:
        print("there are no orders")
        return
    print(f"total orders: {len(orders_track)}")
    for j, o in enumerate(orders_track, 1):
        print(f"  {j}. {o['name']} x{o['quantity']}, euro{o['price']:.2f}, {o['status']}")

def total_revenue(cat):
    r = 0
    for o in orders_track:
        if o['status'] != 'completed':
            continue
        if cat is None or o['category'] == cat:
            r += o['price']
    lbl = cat if cat else "all"
    print(f"revenue ({lbl}): euro{r}")
    return r

def popular_products():
    sold = {}
    
    for o in orders_track:
        nome_corrente = o['name']
        trovato = False
        for nome_in_dizionario in sold:
            if nome_in_dizionario == nome_corrente:
                sold[nome_in_dizionario] += o['quantity']
                trovato = True
                break
        if not trovato:
            sold[nome_corrente] = o['quantity']

    if not sold:
        print("no orders yet")
        return

    a = []
    for n in sold:
        a.append([n, sold[n]])

    n_elementi = len(a)
    for i in range(n_elementi):
        for k in range(0, n_elementi - i - 1):
            if a[k][1] < a[k + 1][1]:

                temp = a[k]
                a[k] = a[k + 1]
                a[k + 1] = temp

    j = 1
    for entry in a:
        n = entry[0]
        q = entry[1]
        print(f"  {j}. {n} - {q} units sold")
        j += 1

def export_report():
    lines = []
    
    count_orders = 0
    for i in orders_track:
        count_orders += 1
        
    lines.append("order summary")
    lines.append("total orders: " + str(count_orders) + "\n")
    

    j = 1
    for o in orders_track:
        riga = str(j) + ". " + o['name'] + " x" + str(o['quantity']) + " | euro" + str(o['price']) + " | " + o['status'] + "\n"
        lines.append(riga)
        j += 1

    lines.append("revenue")
    
    r = 0
    for o in orders_track:
        if o['status'] == 'completed':
            r = r + o['price']
    lines.append("total: euro" + str(r) + "\n")

    lines.append("popular products")
    
    sold = {}
    for o in orders_track:
        nome = o['name']
        if nome in sold:
            sold[nome] = sold[nome] + o['quantity']
        else:
            sold[nome] = o['quantity']
            
    lista_popolari = []
    for chiave in sold:
        lista_popolari.append([chiave, sold[chiave]])
        
    n_pop = 0
    for _ in lista_popolari: n_pop += 1
    
    for i in range(n_pop):
        for k in range(0, n_pop - i - 1):
            if lista_popolari[k][1] < lista_popolari[k+1][1]:
                temp = lista_popolari[k]
                lista_popolari[k] = lista_popolari[k+1]
                lista_popolari[k+1] = temp
                
    contatore_pop = 1
    for elem in lista_popolari:
        n = elem[0]
        q = elem[1]
        lines.append(str(contatore_pop) + ". " + n + " - " + str(q) + " units\n")
        contatore_pop += 1

    f = open('report.txt', 'w')
    for riga_da_scrivere in lines:
        f.write(riga_da_scrivere)
    f.close()
    
    print("saved to report.txt")


# Menu

def menu():
    while True:
        print("\n menu")
        print("1. browse products")
        print("2. place an order")
        print("3. biew reports")
        print("4. quit")

        choice = input("choose an option: ")

        if choice == "1":
            cat = input("category (blank = all): ")
            if cat:
                show_by_category(cat)
            else:
                show_all()

        elif choice == "2":
            prod = input("product name: ")
            try:
                q = int(input("quantity: "))
            except ValueError:
                print("invalid")
                continue
            place_order(prod, q)

        elif choice == "3":
            print("\n  1. order summary")
            print("  2. total revenue")
            print("  3. popular products")
            print("  4. export report")
            sub = input("choose: ")
            if sub == "1":
                order_summary()
            elif sub == "2":
                cat = input("category (blank = all): ")
                total_revenue(cat)
            elif sub == "3":
                popular_products()
            elif sub == "4":
                export_report()

        elif choice == "4":
            print("exited")
            break

        else:
            print("invalid")


menu()
```

```python
# 18

#!/bin/python3

import math
import os
import random
import re
import sys



#
# Complete the 'kmer_counting' function below.
#
# The function is expected to return a 2D_STRING_ARRAY.
# The function accepts following parameters:
#  1. STRING string
#  2. INTEGER k
#

def kmer_counting(string, k):
    
    dict = {}
    j = []
    l = list(string)
    s = 0
    e = k
    for i in range (len(l) -k +1):
        
        sl = string[s : e]
        j.append(sl)
        s += 1
        e +=1
        dict.update({sl:0})
    
    for key in dict.keys():
        
        n = j.count(key)
        dict[key] = n
    result = []
    
    for key, value in dict.items():
        result.append([key, value])
        
    result.sort()

    return result
    
    

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    string = input()

    k = int(input().strip())

    result = kmer_counting(string, k)

    fptr.write('\n'.join(["%s %d"%(x[0], x[1]) for x in result]))
    fptr.write('\n')

    fptr.close()
```

```python
# 16

#!/bin/python3

import math
import os
import random
import re
import sys



#
# Complete the 'sort_remove' function below.
#
# The function is expected to return an INTEGER_ARRAY.
# The function accepts following parameters:
#  1. INTEGER_ARRAY list
#  2. INTEGER elem
#  3. INTEGER k
#

def sort_remove(l, elem, n):
    result = []
    for i in l:
        if i == elem:
            l.remove(i)
        else:
            continue
    l.sort(reverse=True)
    result = l[:n]
    
    
    return result

if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    list_count = int(input().strip())

    l = []

    for _ in range(list_count):
        list_item = int(input().strip())
        l.append(list_item)

    elem = int(input().strip())

    n = int(input().strip())

    result = sort_remove(l, elem, n)

    fptr.write('\n'.join(map(str, result)))
    fptr.write('\n')

    fptr.close()
```

```python
# 14

#!/bin/python3

import math
import os
import random
import re
import sys


# Complete the 'linspace' function below.
# The function is expected to return a FLOAT_ARRAY.
# The function accepts following parameters:
#  1. FLOAT start
#  2. FLOAT stop
#  3. INTEGER num
def linspace(start, stop, num):
    
    step = (stop - start) / num
    l = []
    for i in range(num):
        l.append(start)
        start += step
        start = round(start, 2)
    return l
    
    
    
    
    
    
if __name__ == '__main__':
    fptr = open(os.environ['OUTPUT_PATH'], 'w')

    start = float(input().strip())

    stop = float(input().strip())

    num = int(input().strip())

    result = linspace(start, stop, num)

    fptr.write('\n'.join(map(str, result)))
    fptr.write('\n')

    fptr.close()
```
